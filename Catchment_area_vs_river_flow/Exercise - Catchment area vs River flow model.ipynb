# Catchment area vs River flow model
This Notebook defines a function to model the relationship between catchment area and river flow using a linear function and calculates the error between the observed data and the model's prediction. It also optionally generates a plot to visualize the data and the model fit.
![](util/area_flow.png)
## Import libraries
As always, first we need to import the necessary libraries.
- **NumPy**: This library provides efficient numerical operations on arrays and matrices, including mathematical functions, linear algebra, and random number generation. It's crucial for performing numerical computations in Python.
- **Matplotlib**: This library is used to create a wide range of static, animated, and interactive visualizations, such as line plots, scatter plots, histograms, and bar charts. It offers a flexible interface for customizing plots and saving them in various formats.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Create the model
This code defines a function called `area_flow` that takes four arguments:

* `observed_area`: This is a NumPy array containing observed catchment areas. 
* `observed_flow`: This is another NumPy array containing observed river flow values corresponding to the catchment areas.
* `a`: This is a floating-point number representing a parameter in the linear model.
* `b`: This is another floating-point number representing another parameter in the linear model.

The function performs the following steps:

1. **Model Prediction:** It calculates the predicted river flow based on a linear model. The model equation is `flow = a * area + b`, where `a` and `b` are the provided parameters, and `area` is the catchment area. It multiplies the `observed_area` array with `a` and adds `b` to get an array of predicted flow values.

2. **Visualization:** It creates a plot to visualize the observed data and the model's prediction. 
    * It uses `plt.scatter` to plot the observed catchment areas (x-axis) vs. observed river flows (y-axis) as blue dots.
    * It uses `plt.plot` to draw a red line representing the model prediction (predicted flow vs. catchment area).

3. **Adding labels and formatting (Optional):** It adds labels to the x-axis (Catchment Area), y-axis (River Flow), and a title (`Catchment Area vs. River Flow`). It also sets limits for the x-axis (0 to 60) and y-axis (0 to 200) and adds a legend to differentiate between the observed data and model prediction lines. Finally, it displays the plot using `plt.show()`.

4. **Error Calculation (Optional):** It calculates the mean squared error (MSE) between the observed flow and the predicted flow. The MSE is a common metric to measure how well a model fits the data. Lower MSE indicates a better fit.

5. **Return Value:** The function returns the calculated error. 

```python
def area_flow(observed_area, observed_flow, a, b, x_min = 5, x_max = 50):
    """
    This function model the relationship between catchment area and river flow as linear (flow = a * area + b)
    
    Args:
        observed_area (np.array): Array of observed catchment areas.
        observed_flow (np.array): Array of observed river flow values.
        a,b (float): Model parameters 
    
    Returns:
        error(float): mean squared error between the observed data and the model result.
    """
    
    # Plot observed data (dots)
    plt.scatter(observed_area, observed_flow, color='blue', label='Observed data')
    
    # Plot model results (line)
    plt.plot(np.arange(x_min,x_max), a * np.arange(x_min,x_max) + b, color='red', label='Model prediction')
    
    # Add labels and legend
    plt.xlabel('Catchment Area')
    plt.ylabel('River Flow')
    plt.title('Catchment Area vs. River Flow')
    plt.xlim(x_min,x_max)
    plt.ylim(0,200)
    plt.legend()
    plt.show()
    
    # Calculate mean squared error (optional)
    predicted_flow = a * observed_area + b # Calculate the model-predicted flow based on the model function and parameters
    error = np.mean((observed_flow - predicted_flow) ** 2)
    print(f"Mean Squared Error: {int(error)}")
    
```

## Apply the model

```python
# Example data
observed_area = np.array([5, 9, 20, 28, 40, 50])
observed_flow = np.array([40,70, 120, 125, 160, 180])

# Define model parameters (a, b)
a = 3
b = 40

# Plot the model and data, and calculate error
error = area_flow(observed_area, observed_flow, a ,b)
```

## Create an alternative model
Instead of a linear model, this model calculates the predicted river flow based on a power-law equation. The model equation is `flow = a * area ** b`, where `a` and `b` are the provided parameters, and `area` is the catchment area. It multiplies the `observed_area` array with `a` and **raises it to the power of**  `b` to get an array of predicted flow values.

```python
def area_flow_alt(observed_area, observed_flow, a, b, x_min = 5, x_max = 50):
    """
    This function model the relationship between catchment area and river flow as a power-law (flow = a * area ** b)
    
    Args:
        observed_area (np.array): Array of observed catchment areas.
        observed_flow (np.array): Array of observed river flow values.
        a,b (float): Model parameters 
    
    Returns:
        error(float): mean squared error between the observed data and the model result.
    """
    
    # Plot observed data (dots)
    plt.scatter(observed_area, observed_flow, color='blue', label='Observed data')
    
    # Plot model results (line)
    plt.plot(np.arange(x_min,x_max), a * np.arange(x_min,x_max) ** b, color='red', label='Model prediction')
    
    # Add labels and legend
    plt.xlabel('Catchment Area')
    plt.ylabel('River Flow')
    plt.title('Catchment Area vs. River Flow')
    plt.xlim(x_min,x_max)
    plt.ylim(0,200)
    plt.legend()
    plt.show()
    
    # Calculate mean squared error (optional)
    predicted_flow = a * observed_area ** b # Calculate the model-predicted flow based on the model function and parameters
    error = np.mean((observed_flow - predicted_flow) ** 2)
    print(f"Mean Squared Error: {int(error)}")
    
    return error
```

## Apply the alternative model

```python
# Define model parameters (a, b)
a = 25
b = 0.50

# Plot the model and data, and calculate error
error = area_flow_alt(observed_area, observed_flow, a ,b)
```
