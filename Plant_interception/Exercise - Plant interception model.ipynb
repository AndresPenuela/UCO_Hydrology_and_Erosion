# Plant interception model - Braden

![](images/plant_interception.jpg)

This Python function, `plant_interception_Braden`, calculates and visualizes plant interception (PI) based on daily rainfall using the **Braden model**. Here's a step-by-step breakdown:

### Parameters:
- `a`: Empirical parameter that influences how plant interception is calculated (units in cm d^-1).
- `LAI`: Leaf Area Index, which is a measure of the leaf surface area relative to the ground surface area.
- `min_rain`: Minimum daily rainfall value (in mm) for plotting.
- `max_rain`: Maximum daily rainfall value (in mm) for plotting.

### Function Logic:
1. **Generate Rainfall Data**:
   - The function creates an array, `rain`, using `np.arange` that contains rainfall values from `min_rain` to `max_rain`. These values represent the range of daily rainfall in millimeters.

2. **Calculate Plant Interception (PI)**:
   - For each rainfall value in `rain`, the function computes the corresponding plant interception (PI) using the Braden model formula:
     $$
     PI = a \times LAI \times \left(1 - \frac{1}{1 + \frac{\text{rain}}{3a}}\right)
     $$
     - The rainfall values are divided by 10 to convert from mm to cm.
     - The formula models how much rainfall is intercepted by plants based on rainfall amount, `a`, and `LAI`.
     - The final result, `PI`, is in millimeters (mm).

3. **Plot the Results**:
   - A plot is created using Matplotlib to show the relationship between daily rainfall (`rain`) and plant interception (`PI`).
   - The x-axis represents daily rainfall in millimeters.
   - The y-axis represents plant interception (PI) in millimeters.
   - The plot includes:
     - Title: "Plant Interception (PI) vs. Daily Rainfall"
     - Labels for the axes.
     - A grid for easier reading.
     - X-axis limits are set between `min_rain` and `max_rain`.
     - Y-axis limits are set from 0 to 40 mm to ensure some extra space above the maximum PI values for better visualization.

Finally, the plot is displayed, helping to visualize how plant interception varies with rainfall.
## Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Define function to represent the model

```python
def plant_interception_Braden(a, LAI, min_rain, max_rain):
    """
    Calculate plant interception (PI) using the Braden model and plot it for a range of rainfall values.

    Parameters:
    a (float): Empirical parameter (cm d^-1)
    LAI (float): Leaf area index
    min_rain (float): Minimum daily rainfall (in mm)
    max_rain (float): Maximum daily rainfall (in mm)

    """
    # Create an array of rain values from min_rain to max_rain
    rain = np.arange(min_rain, max_rain) # mm

    # Calculate plant interception for each rain value (converted to mm)
    PI = a * LAI * (1 - 1 / (1 + rain/10 / (3 * a))) * 10 # rain_values are converted from mm to cm

    # Plotting the results
    plt.figure(figsize=(6, 4))
    plt.plot(rain, PI, color='blue', linewidth=2)
    plt.title('Plant Interception (PI) vs. Daily Rainfall')
    plt.xlabel('Daily Rainfall (mm)')
    plt.ylabel('Plant Interception (PI) (mm)')
    plt.grid()
    plt.xlim(min_rain, max_rain)
    plt.ylim(0, 40)  # Add some space above the maximum PI value
    plt.show()
```

## Apply the model

```python
# Example usage
a = 0.5  # Example empirical parameter in cm d^-1
LAI = 3.0  # Example leaf area index
min_rain = 10 # Minimum daily rainfall in mm
max_rain = 60  # Maximum daily rainfall in mm

# Call the function and get the PI values
plant_interception_Braden(a, LAI, min_rain, max_rain)
```
